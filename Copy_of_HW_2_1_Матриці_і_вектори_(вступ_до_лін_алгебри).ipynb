3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
result_matrix = B @ A   # це мені LLM сказав, що є такий оператор матричного множення "@""
print(result_matrix)
result_matrix.shape
```

```python
result_matrix = np.dot(B, A) # матричне множення, як в лекційному ноутбуці ;)
print(result_matrix)
result_matrix.shape
```

2. Отримаємо розмірність (2, 1)

```python
result_vector = result_matrix @ C
print(result_vector)
result_vector.shape
```

3. Отримаємо матрицю розмірністю (2, 2)

```python
result_matrix_D = result_matrix @ D
print(result_matrix_D)
result_matrix_D.shape
```
