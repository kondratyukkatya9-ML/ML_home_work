1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sp.Symbol('x')
f = 10*x - 5
df = sp.diff(f, x)
print(df)
```

```python
x = np.linspace(0, 10, 100)
f = 10*x - 5
df = np.full_like(x, 10)

plt.plot(x, f, label='f(x) = 10x - 5', color='blue')
plt.plot(x, df, label="f'(x) = 10", color='green')
plt.title("Оцінка від кількості годин навчання")
plt.xlabel("Години навчання")
plt.ylabel("Оцінка")
plt.legend()
plt.grid(True)
plt.show()
```

Оскільки наша похідна є константою і довівнює 10, то з кожною додатковою годиною навчання, отримана оцінка зростатиме на 10 балів


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.Symbol('x')
f = x**2 - 5*x + 1
df = sp.diff(f, x)
print(df)
```

```python
x = np.linspace(0, 100, 100)
f = x**2 - 5*x + 1
df = 2*x - 5

plt.plot(x, f, label='f(x) = x² - 5x + 1', color='blue')
plt.plot(x, df, label="f'(x) = 2x - 5", color='green')
plt.title("Витрати компанії від обсягу виробництва")
plt.xlabel("Кількість одиниць")
plt.ylabel("Витрати")
plt.legend()
plt.grid(True)
plt.show()
```
